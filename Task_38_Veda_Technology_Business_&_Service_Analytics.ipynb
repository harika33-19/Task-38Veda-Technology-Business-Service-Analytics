```python
daily = clean_df.groupby("Date")[
    "Inquiries"
].sum().reset_index()

daily = daily.sort_values("Date")

daily = daily.set_index("Date").asfreq("D")

daily["Inquiries"] = daily["Inquiries"].fillna(0)

display(daily.head())
```

```python
daily["Previous_Day"] = daily["Inquiries"].shift(1)

daily["Previous_7_Days"] = (
    daily["Inquiries"].shift(1).rolling(7).mean()
)

daily["Previous_30_Days"] = (
    daily["Inquiries"].shift(1).rolling(30).mean()
)

daily = daily.dropna()
```

```python
X = daily[
    ["Previous_Day", "Previous_7_Days", "Previous_30_Days"]
]

y = daily["Inquiries"]

split = int(len(daily) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Model trained successfully!")
```

```python
mae = mean_absolute_error(y_test, predictions)

r2 = r2_score(y_test, predictions)

print("Mean Absolute Error:", mae)

print("R2 Score:", r2)
```

```python
np.random.seed(42)

n = 1000

services = np.random.choice(
    ["Internship", "Training", "IT Services", "Digital Services"],
    n
)

visitors = np.random.randint(50, 501, n)

inquiries = np.array([
    np.random.binomial(v, 0.10) for v in visitors
])

applications = np.where(
    services == "Internship",
    np.random.binomial(inquiries, 0.45),
    0
)

training_enrollments = np.where(
    services == "Training",
    np.random.binomial(inquiries, 0.35),
    0
)

conversions = np.random.binomial(inquiries, 0.30)

revenue = conversions * np.random.randint(1000, 5001, n)

df = pd.DataFrame({
    "Record_ID": range(1, n + 1),
    "Date": np.random.choice(
        pd.date_range("2025-01-01", "2026-10-31"),
        n
    ),
    "Customer_ID": np.random.randint(1001, 1301, n),
    "Region": np.random.choice(
        ["Andhra Pradesh", "Telangana", "Karnataka", "Tamil Nadu"],
        n
    ),
    "Marketing_Channel": np.random.choice(
        ["Google", "Instagram", "LinkedIn", "YouTube", "Referral"],
        n
    ),
    "Service": services,
    "Visitors": visitors,
    "Inquiries": inquiries,
    "Applications": applications,
    "Training_Enrollments": training_enrollments,
    "Conversions": conversions,
    "Revenue": revenue,
    "Customer_Satisfaction": np.random.randint(1, 6, n),
    "Response_Time_Hours": np.random.randint(1, 49, n)
})

df["Date"] = pd.to_datetime(df["Date"])

print("Dataset generated successfully!")
display(df.head(10))
```
